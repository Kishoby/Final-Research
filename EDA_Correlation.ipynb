```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
path="/content/drive/MyDrive/Research/16 months.csv"
df = pd.read_csv(path)

# Define target column
target = "air_quality_PM2.5"

# Select ONLY numeric columns (to avoid errors)
numeric_df = df.select_dtypes(include=['number'])

# Check if target is numeric
if target not in numeric_df.columns:
    print(f"ERROR: Target '{target}' is not numeric or not found in dataset.")
else:
    # Compute correlation in percentage
    corr_percent = numeric_df.corr()[target].sort_values(ascending=False) * 100

    print("\n✅ Correlation (%) of ALL numeric features with PM2.5:\n")
    print(corr_percent)

    # Heatmap for ALL numeric features
    plt.figure(figsize=(12,10))
    sns.heatmap(numeric_df.corr(), annot=False, cmap="coolwarm")
    plt.title("Correlation Heatmap (All Numeric Features)")
    plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
path="/content/drive/MyDrive/Research/16 months.csv"
df = pd.read_csv(path)

# Define the features to check
features = [
     "air_quality_us-epa-index",
    "air_quality_gb-defra-index",
    "air_quality_PM10",
    "air_quality_Carbon_Monoxide",
    "air_quality_Nitrogen_dioxide",
    "air_quality_Sulphur_dioxide",
    "humidity",
    "cloud",
    "visibility_miles",
    "visibility_km",
    "longitude",
    "gust_kph",        # or gust_mph
    "precip_mm",
    "precip_in",
    "air_quality_Ozone"
]

# Include the target
target = "air_quality_PM2.5"
cols_to_check = features + [target]

# Select only numeric columns from the chosen ones
numeric_df = df[cols_to_check].select_dtypes(include=['float64', 'int64'])

# Correlation with target
correlation = numeric_df.corr()[target].sort_values(ascending=False)
print(f"\nCorrelation of selected features with target ({target}):")
print(correlation)

# Heatmap for visualization
plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title(f"Correlation Heatmap with Target: {target}")
plt.show()
```
